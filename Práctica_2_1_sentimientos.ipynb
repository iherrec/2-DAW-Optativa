# **Qué es el análisis de sentimiento**

Es una tarea de clasificación de texto. Dado un texto, el modelo decide si la intención utilizada es positiva, negativa o neutral.

No es "**entender**" el texto como lo haría una persona. Es detectar patrones de palabras y contexto que, estadísticamente, se asocian a cada categoría.

Para esta práctica en español, el modelo más adecuado es **RoBERTuito**. Es un modelo entrenado específicamente con **500 millones de tweets en español**, lo que le permite entender jerga, emojis, abreviaturas y texto informal.

Está diseñado para redes sociales, que es donde el análisis de sentimiento tiene más aplicación real. Y funciona bien en tareas de sentimiento, odio e ironía en español.

RoBERTuito no se entrenó desde cero. Es un modelo preentrenado: alguien invirtió muchísimo tiempo y recursos en entrenarlo con textos masivos para que "entienda" el español.

Luego ese modelo base se ajustó (mediante fine-tuning) con un conjunto más pequeño de tweets etiquetados como positivos, negativos o neutros. El resultado es un modelo especializado en análisis de sentimiento en español, listo para usar.

Esto es importante porque explica por qué tú no entrenas nada. Usas un modelo que ya existe, ya está entrenado, y solo lo llamas desde Colab. Es exactamente lo que haría un desarrollador profesional: **no reinventar la rueda, usar lo que ya funciona**.

# **Hugging Face**

Hugging Face es la plataforma donde viven estos modelos. Su función **pipeline** es una abstracción: tú le dices "quiero realizar un análisis de sentimiento" y "usa este modelo", y ella se encarga de todo lo demás (preprocesar el texto, pasarlo por el modelo, devolver el resultado).

Solo se necesita entender que pipeline es el "botón" que conecta el texto con un modelo ya entrenado.

El modelo devuelve dos cosas:

    Una etiqueta: POS, NEG o NEU

    Una confianza (score), un número entre 0 y 1

La confianza no es lo mismo que **"acierto"**. Es la seguridad del modelo en su propia predicción. Un score de 0.99 no garantiza que tenga razón; un score de 0.51 en una frase ambigua es normal y no es un fallo.

El modelo no lee intenciones, lee patrones. El guiño o el signo de interrogación alteran el contexto que el modelo aprendió durante el preentrenamiento, y eso mueve la predicción.

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")
# carga un modelo de inteligencia artificial
# entrenado para analizar el sentimiento
# de textos en español, con especial enfoque
# en lenguaje informal y redes sociales

# === este es el texto  que se clasifica ===
texto = "Me encanta este módulo de IA"
# =====================================

resultado = clasificador(texto)
print(resultado)
```

```python
# clasifica un texto que sea negativo

texto = "Estoy harto de todo."
resultado = clasificador(texto)
print(resultado)
```

```python
# clasifica un texto neutro
texto = "Son las 13:37"
resultado = clasificador(texto)
print(resultado)
```

```python
# clasifica un texto irónico
texto = "Qué bien, solo han tardado 2 horas en atenderme en el bar."
texto2 = "Qué pena, me he encontrado 50€ en la calle"
resultado = clasificador(texto)
resultado2 = clasificador(texto2)
print(resultado)
print(resultado2)
```

```python
# clasifica un texto usando emojis
textos = [
    "Estoy enfadado 😂",
    "Estoy enfadado 🤬",
    "Estoy feliz 😨",
    "Estoy feliz 😊"
]

resultados = clasificador(textos)

for texto, res in zip(textos, resultados):
    print(f"'{texto}' -> Label: {res['label']} | Score: {res['score']:.2f}")
```

# **Práctica**

Implementa en python un analizador de comentarios. Los comentarios positivos y nuetros serán aceptados, salvo que contengan alguna palabra de la lista negra de palabras. Los comentarios negativos que contengan alguna palabra de la lista negra o que tengan un score de más de 0.5 serán rechazadas.

```python
# pon aquí el código de tu analizador
# El analizador hará un análisis de unas frases ya cargadas, pero también da la posibilidad al usuario de introducir una frase.

# 1. Instalación de la librería necesaria (descomenta si no la tienes instalada en Colab)
# !pip install -q transformers torch

from transformers import pipeline

# 2. Definición de la lista negra de palabras
palabras_baneadas = {
    "pendejo", "mierda", "nazi", "hitler", "subnormal",
    "gilipollas", "cabrón", "puta", "polla", "maricón"
}

# 3. Carga del modelo RobertUITO para análisis de sentimiento
clasificador = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")

# 4. Función de evaluación para reutilizar el código
def evaluar_frase(frase):
    frase_lower = frase.lower()

    # Comprobación 1: Contiene alguna palabra de la lista negra
    contiene_palabra_baneada = any(palabra in frase_lower for palabra in palabras_baneadas)

    # Comprobación 2: Análisis de sentimiento con RobertUITO
    resultado = clasificador(frase)[0]
    label = resultado['label']   # 'POS', 'NEU' o 'NEG'
    score = resultado['score']   # Confianza del modelo (0.0 a 1.0)

    # Criterio de rechazo: contiene palabra de la lista negra OR (es NEG y score > 0.50)
    es_negativo_alto = (label == 'NEG' and score > 0.50)

    es_no_permitida = contiene_palabra_baneada or es_negativo_alto

    razones = []
    if contiene_palabra_baneada:
        razones.append("Lista negra")
    if es_negativo_alto:
        razones.append(f"Sentimiento NEG (Score: {score:.2f})")

    return es_no_permitida, label, score, ", ".join(razones)

# ---------------------------------------------------------
# PROCESAMIENTO DE LA COLECCIÓN DE FRASES
# ---------------------------------------------------------
frases = [
    "¡Qué bonito día hace hoy para pasear!",
    "Eres un completo gilipollas por hacer eso.",
    "No me gusta nada esta comida, sabe fatal.",
    "Ese tipo es un nazi de cuidado.",
    "Buen vídeo, pero has cometido varios errores. Intenta mejorar para la próxima.",
    "Vaya mierda de película, me pareció horriblemente aburrida y pésima.",
    "Tengo muchas ganas de ver la nueva temporada.",
    "No me da la puta gana de hacer los deberes.",
    "Hoy para comer tengo macarrones con tomate",
    "Una polla es una gallina adulta.",

]

frases_no_permitidas = []
frases_aprobadas = []

for frase in frases:
    no_permitida, label, score, razon = evaluar_frase(frase)
    if no_permitida:
        frases_no_permitidas.append((frase, razon))
    else:
        frases_aprobadas.append((frase, label, score))

print("--- RESULTADOS DE LA COLECCIÓN INICIAL ---")
print("\n❌ FRASES NO PERMITIDAS:")
for frase, razon in frases_no_permitidas:
    print(f" - \"{frase}\" -> [Motivo: {razon}]")

print("\n✅ FRASES APROBADAS:")
for frase, label, score in frases_aprobadas:
    print(f" - \"{frase}\" -> [{label} - Score: {score:.2f}]")

# ---------------------------------------------------------
# NUEVA SECCIÓN: ENTRADA DEL USUARIO
# ---------------------------------------------------------
print("\n" + "="*50)
print("--- EVALUAR FRASE PERSONALIZADA ---")

# Solicitar frase al usuario mediante un campo de texto en Colab
frase_usuario = input("Introduce una frase para clasificar: ")

if frase_usuario.strip():  # Comprobar que no esté vacía
    no_permitida, label, score, razon = evaluar_frase(frase_usuario)

    print("\nRESULTADO PARA TU FRASE:")
    if no_permitida:
        print(f"❌ La frase NO HA SIDO PERMITIDA.")
        print(f"   Motivo: {razon}")
    else:
        print(f"✅ La frase HA SIDO APROBADA.")
        print(f"   Clasificación: {label} (Confianza: {score:.2f})")
else:
    print("No has introducido ninguna frase.")
```
